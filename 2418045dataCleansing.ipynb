##3.Data Cleansing – Standarisasi
Lalu Tahap ini dilakukan standarisasi untuk menyamakan format data yang masih memiliki perbedaan penulisan. Tujuannya agar data menjadi lebih konsisten dan mudah digunakan untuk proses analisis

###3.1 Standarisasi SHARE_CODE

Kode saham seharusnya memiliki format yang konsisten. Pada tahap ini  membersihkan spasi dan menyamakan huruf menjadi huruf kapita

```python
df['SHARE_CODE'] = (
    df['SHARE_CODE']
    .astype('string')
    .str.strip()
    .str.upper()
)

df[['SHARE_CODE']].head()
```

###3.2 Standarisasi ISSUER_NAME

Kolom ISSUER_NAME dibersihkan dengan menghapus spasi berlebih dan menyamakan format penulisan menjadi huruf kapital

```python
df['ISSUER_NAME'] = (
    df['ISSUER_NAME']
    .astype('string')
    .str.strip()
    .str.upper()
)

df[['ISSUER_NAME']].head()
```

###3.3 Standarisasi INVESTOR_NAME

Kolom INVESTOR_NAME dibersihkan dengan menghapus spasi yang tidak diperlukan dan menyamakan penulisan menjadi huruf kapital

```python
df['INVESTOR_NAME'] = (
    df['INVESTOR_NAME']
    .astype('string')
    .str.strip()
    .str.upper()
)

df[['INVESTOR_NAME']].head()
```

```python
df['INVESTOR_CLASSIFICATION'] = (
    df['INVESTOR_CLASSIFICATION']
    .astype('string')
    .str.strip()
    .str.upper()
)

df[['INVESTOR_CLASSIFICATION']].head()
```

###3.4 Standarisasi INVESTOR_CLASSIFICATION
Pada kolom INVESTOR_CLASSIFICATION, dilakukan standarisasi dengan menghilangkan spasi berlebih dan mengubah seluruh huruf menjadi huruf kapital agar format data menjadi konsisten.

###3.5 Standarisasi DATE

Standarisasi tanggal dilakukan agar format tanggal menjadi seragam dan lebih mudah digunakan dalam analisis

```python
df['DATE'] = pd.to_datetime(
    df['DATE'],
    errors='coerce',
    dayfirst=True
)

df['DATE'] = df['DATE'].dt.strftime('%Y-%m-%d')

df[['DATE']].head()
```

###3.6 Standarisasi LOCAL_FOREIGN

Kolom LOCAL_FOREIGN distandarisasi dengan menghapus spasi yang tidak diperlukan dan menyamakan penulisan menjadi huruf kapital

```python
# Membersihkan spasi dan menyamakan huruf
df['LOCAL_FOREIGN'] = (
    df['LOCAL_FOREIGN']
    .astype('string')
    .str.strip()
    .str.upper()
)

# Menyamakan kode L/F dengan LOCAL/FOREIGN
df['LOCAL_FOREIGN'] = df['LOCAL_FOREIGN'].replace({
    'L': 'LOCAL',
    'F': 'FOREIGN'
})

df[['LOCAL_FOREIGN']].value_counts()
```

###3.7 Standarisasi Data Numerik

Kolom jumlah kepemilikan saham dikonversi menjadi tipe data numerik, Karakter pemisah angka yang tidak diperlukan akan dibersihkan sebelum data dikonversi

```python
numeric_columns = [
    'HOLDINGS_SCRIPLESS',
    'HOLDINGS_SCRIP',
    'TOTAL_HOLDING_SHARES'
]

for col in numeric_columns:
    df[col] = (
        df[col]
        .astype('string')
        .str.replace(',', '', regex=False)
        .str.replace('.', '', regex=False)
        .str.strip()
    )

    df[col] = pd.to_numeric(
        df[col],
        errors='coerce'
    )

display(
    df[numeric_columns].style.format('{:,.0f}')
)
```

###3.8 Standarisasi PERCENTAGE

Kolom PERCENTAGE dibersihkan dari simbol persen dan disesuaikan format desimalnya, kemudian dikonversi menjadi tipe data numerik

```python
df['PERCENTAGE'] = (
    df['PERCENTAGE']
    .astype('string')
    .str.replace('%', '', regex=False)
    .str.replace(',', '.', regex=False)
    .str.strip()
)

df['PERCENTAGE'] = pd.to_numeric(
    df['PERCENTAGE'],
    errors='coerce'
)

df[['PERCENTAGE']].head()
```

## 4.Data Cleansing – Missing Value
Tahap ini dilakukan penanganan terhadap data yang kosong. Data kosong diperiksa terlebih dahulu, kemudian diberikan nilai yang sesuai agar dataset dapat digunakan untuk analisi

###4.1 Identifikasi Missing Value

```python
missing_value = df.isna().sum()

print("Jumlah missing value:")
print(missing_value)
```

###4.2 Penanganan Missing Value

Untuk data kategorikal yang kosong, gunakan nilai Unknown karena nilai sebenarnya tidak diketahui. agar tidak mengarang nilai yang seharusnya

```python
categorical_columns = [
    'INVESTOR_CLASSIFICATION',
    'LOCAL_FOREIGN',
    'NATIONALITY',
    'DOMICILE'
]

for col in categorical_columns:
    df[col] = df[col].fillna('Unknown')

print("Missing value setelah penanganan:")
print(df.isna().sum())
```

## 5.Data Cleansing – Deduplikasi
Data duplikat perlu diperiksa karena keberadaan data yang sama dapat memengaruhi hasil analisis

###5.1 Identifikasi Data Duplikat

```python
duplicate_count = df.duplicated().sum()

print("Jumlah data duplikat:", duplicate_count)
```

## 6.Data Enrichment
Pada tahap ini dilakukan penambahan informasi berdasarkan data yang tersedia melalui tabel referensi (lookup), Informasi tambahan digunakan untuk membuat data lebih mudah dipahami

###6.1 Membuat Data Lookup

```python
lookup_investor = pd.DataFrame({
    'LOCAL_FOREIGN': ['LOCAL', 'FOREIGN', 'UNKNOWN'],
    'INVESTOR_TYPE': [
        'Investor Lokal',
        'Investor Asing',
        'Tidak Diketahui'
    ]
})

lookup_investor
```

###6.2 Menggabungkan Data Lookup

Sebelum melakukan merge, pastikan nilai LOCAL_FOREIGN yang kosong sudah menjadi Unknown dan disesuaikan dengan lookup

```python
df['LOCAL_FOREIGN'] = (
    df['LOCAL_FOREIGN']
    .fillna('UNKNOWN')
    .astype('string')
    .str.strip()
    .str.upper()
)

df['LOCAL_FOREIGN'] = df['LOCAL_FOREIGN'].replace({
    'L': 'LOCAL',
    'F': 'FOREIGN'
})
```

```python
df['INVESTOR_TYPE'] = df['INVESTOR_TYPE'].fillna('Tidak Diketahui')

df[['LOCAL_FOREIGN', 'INVESTOR_TYPE']].head(10)
```
